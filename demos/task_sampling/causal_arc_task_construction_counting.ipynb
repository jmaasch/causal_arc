```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import colors
import seaborn as sns
import json
import platform
import ast
from itertools import permutations,product
import os

# Custom modules.
os.chdir("../causal_arc")
from carc_utils import UtilsARC
from carc_augment import AugmentARC
from carc import CausalARC
from carc_tasks_counting import TaskCounting

# View versioning.
print("python version     :", platform.python_version())
print("numpy version      :", np.__version__)
print("pandas version     :", pd.__version__)
print("matplotlib version :", matplotlib.__version__)
print("seaborn version    :", sns.__version__)
```

```python
u = UtilsARC()
a = AugmentARC()
c = CausalARC()
t = TaskCounting()
```

```python
# Dict to export.
ttt_dict = dict()

# Total demonstration pairs.
n_examples = 5

# Output grid sizes.
sizes = [(10,10), (15,15), (20,20)]
```

# task_SCMm5ob

```python
for size in sizes:
    task_name = "task_SCMm5ob".split("_")[1]
    task_name += f"_{size[0]}x{size[1]}"
    print(task_name)
    tdict = t.task_SCMm5ob(size = size,
                             n_examples = n_examples, 
                             plot = True,
                             plot_type = "input_output", 
                             figsize = (6,2),
                             grid = True)
    ttt_dict[task_name] = tdict
```
