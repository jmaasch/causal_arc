```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import colors
import seaborn as sns
import json
import platform
import ast
from itertools import permutations,product
import os

# Custom modules.
os.chdir("../causal_arc")
from carc import CausalARC
from carc_utils import UtilsARC
from carc_augment import AugmentARC
from carc_tasks_order import TaskOrder

# View versioning.
print("python version     :", platform.python_version())
print("numpy version      :", np.__version__)
print("pandas version     :", pd.__version__)
print("matplotlib version :", matplotlib.__version__)
print("seaborn version    :", sns.__version__)
```

```python
u = UtilsARC()
a = AugmentARC()
c = CausalARC()
t = TaskOrder()
```

```python
# Dict to export.
ttt_dict = dict()

# Total demonstration pairs.
n_examples = 5
```

# task_SCMv4bg

```python
sizes = [15, 30]
task_name = "SCMv4bg"
for size in sizes:
    tdict = t.task_SCMv4bg(size = size,
                             n_examples = n_examples, 
                             plot = True,
                             plot_type = "input_output", 
                             figsize = (6,2),
                             grid = True)
    ttt_dict[f"{task_name}!{size}x9"] = tdict
```

# task_SCMye7g

```python
#sizes = [(15,15), (20,20), (25,25)]
sizes = [15,20,25]
task_name = "SCMye7g"
for size in sizes:
    tdict = t.task_SCMye7g(size = size,
                             n_examples = n_examples, 
                             plot = True,
                             plot_type = "input_output", 
                             figsize = (6,2),
                             grid = True)
    ttt_dict[f"{task_name}!{size}x{size}"] = tdict
```

# task_SCMswcs

```python
#sizes = [(15,15), (20,20), (25,25)]
task_name = "SCMswcs"
tdict = t.task_SCMswcs(n_examples = n_examples, 
                         plot = True,
                         plot_type = "input_output", 
                         figsize = (6,2),
                         grid = True)
ttt_dict[task_name] = tdict
```

# task_SCMtzlq

```python
sizes = [(20,20), (25,25), (30,30)]
n_sprites_list = [7,8,9]
task_name = "SCMtzlq"
for size,n_sprites in zip(sizes,n_sprites_list):
    tdict = t.task_SCMtzlq(size = size,
                           n_sprites = n_sprites,
                             n_examples = n_examples, 
                             plot = True,
                             plot_type = "input_output", 
                             figsize = (6,2),
                             grid = True)
    ttt_dict[f"{task_name}!{size[0]}x{size[1]}"] = tdict
```

# task_SCMffb8

```python
task_name = "SCMffb8"
tdict = t.task_SCMffb8(n_examples = n_examples, 
                         plot = True,
                         plot_type = "input_output", 
                         figsize = (6,2),
                         grid = True)
ttt_dict[task_name] = tdict
```

# Export JSON

```python
with open('../data/causal_arc/causal_arc_ordering.json', 'w') as f:
    json.dump(ttt_dict, f)
```

```python
# Export solutions file.
ttt_dict_solutions = dict()
for task,d in ttt_dict.items():
    test_output = d["test"][0].get("output")
    ttt_dict_solutions[task] = [test_output]

with open('../data/causal_arc/causal_arc_ordering_solutions.json', 'w') as f:
    json.dump(ttt_dict_solutions, f)
```
