```python
# Dict to export.
ttt_dict = dict()

# Total demonstration pairs.
n_examples = 5
```

# task_SCMrwf3

```python
task_name = "task_SCMfwpq".split("_")[1]
print(task_name)
colors_list = [[6,7,4], [3,2,1]]
sizes = [(20,20)]
for size,colors in zip(sizes,colors_list):
    tdict = t.task_SCMfwpq(size = size,
                           colors = colors,
                           n_examples = n_examples,
                           plot = True,
                           plot_type = "input_output", 
                           figsize = (6,2),
                           grid = True)
    ttt_dict[f"{task_name}!{size[0]}x{size[1]}"] = tdict
```
