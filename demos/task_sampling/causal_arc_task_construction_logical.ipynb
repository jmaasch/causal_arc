```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import colors
import seaborn as sns
import json
import platform
import ast
from itertools import permutations,product
import os

# Custom modules.
os.chdir("../causal_arc")
from carc_utils import UtilsARC
from carc_augment import AugmentARC
from carc import CausalARC
from carc_tasks_logical import TaskLogical

# View versioning.
print("python version     :", platform.python_version())
print("numpy version      :", np.__version__)
print("pandas version     :", pd.__version__)
print("matplotlib version :", matplotlib.__version__)
print("seaborn version    :", sns.__version__)
```

```python
u = UtilsARC()
a = AugmentARC()
c = CausalARC()
tl = TaskLogical()
```

```python
# Dict to export.
ttt_dict = dict()

# Total demonstration pairs.
n_examples = 5

# Output grid sizes.
sizes = [(10,10), (15,15), (20,20)]
```

# Logical and

```python
and_dict = dict()
```

```python
colors_list = [(2,6,7), (4,5,6), (9,8,1), (8,7,3)]
fun = "and"
for size,colors in zip(sizes,colors_list):
    tdict = tl.task_SCMdky5(fun = fun,
                             upper_color = colors[0],
                             lower_color = colors[1],
                             output_color = colors[2],
                             size = size,
                             n_examples = n_examples, 
                             plot = True,
                             plot_type = "input_output", 
                             figsize = (6,2),
                             grid = True)
    ttt_dict[f"SCMdky5{fun}{size[0]}x{size[1]}"] = tdict
    and_dict[f"SCMdky5{fun}{size[0]}x{size[1]}"] = tdict
```

# Alternating rows

```python
alternate_dict = dict()
```

```python
colors_list = [(1,2,3), (4,5,6), (7,8,9), (2,4,6)]
sizes = [(4,4), (8,8)]
fun_0 = "and"
fun_1 = "xor"
for size,colors in zip(sizes,colors_list):
    tdict = tl.task_SCMu3am(size = size, 
                            fun_0 = fun_0,
                            fun_1 = fun_1,
                            alternate_axis = 0, # 1 for columns, 0 for rows.
                            upper_color = colors[0], 
                            lower_color = colors[1],
                            output_color = colors[2],
                            n_examples = n_examples, 
                            plot = True,
                            plot_type = "input_output", 
                            figsize = (6,2),
                            grid = True)
    ttt_dict[f"SCMu3am!{fun_0}!{fun_1}!axis0!{size[0]}x{size[1]}"] = tdict
    alternate_dict[f"SCMu3am!{fun_0}!{fun_1}!axis0!{size[0]}x{size[1]}"] = tdict
```
